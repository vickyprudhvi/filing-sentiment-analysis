```python
import os
import requests
import pandas as pd
```

```python
my_dict ={"User-Agent":"Prudhvi Raj vicky.prudhvi@gmail.com"}
URL='https://www.sec.gov/files/company_tickers.json'
```

```python
response= requests.get(URL,headers=my_dict)
```

```python
response.status_code
```

```python
data =response.json()
```

```python
list(data.items())[:3]
```

```python
data_df= pd.DataFrame(data.values())
```

```python
data_df.head()
```

```python
my_tickers=["NVDA","AAPL","GOOGL","MSFT","AMZN","T","COST","TSLA","UNH","CF"]
```

```python
companies=data_df[data_df["ticker"].isin(my_tickers)].reset_index(drop=True)
companies
```

```python
companies["cik"]=companies["cik_str"].astype(str).str.zfill(10)
companies
```

```python
companies.to_csv("companies.csv",index=False)
```

```python
apple_cik=companies[companies["ticker"]=="AAPL"].iloc[0]["cik"]
apple_cik
```

```python
URL=f"https://data.sec.gov/submissions/CIK{apple_cik}.json"
URL
```

```python
response=requests.get(URL,headers=my_dict)
response.status_code
```

```python
apple=response.json()
```

```python
apple.keys()
```

```python
apple_filings=pd.DataFrame(apple["filings"]["recent"])
apple_filings.head()
```

```python
apple_filings.shape
```

```python
apple_filings["form"].value_counts()
```

```python
apple_8k=apple_filings[apple_filings["form"]=="8-K"]
apple_8k.head()
```

```python
apple_8k["items"].value_counts()
```

```python
apple_earnings=apple_8k[apple_8k["items"].str.contains("2.02",regex=False)]
apple_earnings=apple_earnings[["accessionNumber","filingDate", "reportDate", "form", "items", "primaryDocument"]].reset_index(drop=True)
apple_earnings
```

```python
apple_earnings.shape
```

```python
print(apple_earnings["filingDate"].min())
print(apple_earnings["filingDate"].max())
```

```python
apple_cik_num=companies[companies["ticker"]=="AAPL"].iloc[0]["cik_str"]
apple_cik_num
```

```python
accession=apple_earnings["accessionNumber"].str.replace("-","").iloc[0]
accession
```

```python
Earning_URL=f"https://www.sec.gov/Archives/edgar/data/{apple_cik_num}/{accession}/index.json"
Earning_URL
```

```python
Earning_folder=requests.get(Earning_URL,headers=my_dict)
Earning_folder.status_code
```

```python
folder=Earning_folder.json()
folder
```

```python
folder_df=pd.DataFrame(folder["directory"]["item"])
folder_df
```

```python
press_release_name=folder_df[folder_df["name"].str.contains("ex99",regex=False)]["name"].iloc[0]
press_release_name
```

```python
New_Earning_URL=f"https://www.sec.gov/Archives/edgar/data/{apple_cik_num}/{accession}/{press_release_name}"
New_Earning_URL
```

```python
pressRelease=requests.get(New_Earning_URL,headers=my_dict)
pressRelease.status_code
```

```python
press_release_html=pressRelease.text
len(press_release_html)
```

```python
press_release_html[:1000]
```

```python
folder_path = "data/raw/AAPL"
os.makedirs(folder_path, exist_ok=True)
file_path=f"{folder_path}/{accession}.htm"
with open(file_path, "w", encoding="utf-8") as f:
    f.write(press_release_html)

print(f"Saved {file_path}")
```
