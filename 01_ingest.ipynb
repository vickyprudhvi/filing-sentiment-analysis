```python
def download_press_release(cik_num,accession_num ,ticker):
    folder_path = f"data/raw/{ticker}"    
    accession = accession_num.replace("-", "")
    file_path = f"{folder_path}/{accession}.htm"
    
    if os.path.exists(file_path):
        print( "skipped")
        return
    # Get the list of files in this filing's folder
    folder_url = f"https://www.sec.gov/Archives/edgar/data/{cik_num}/{accession}/index.json"
    # print(f"https://www.sec.gov/Archives/edgar/data/{cik_num}/{accession}")
    folder_response = requests.get(folder_url, headers=my_dict)
    time.sleep(0.2)
    if folder_response.status_code != 200:
        print(f"Failed folder {accession_num}: {folder_response.status_code}")
        return

    folder_df = pd.DataFrame(folder_response.json()["directory"]["item"])

    # Find the exhibit 99 file (the press release)
    matches = folder_df[folder_df["name"].str.contains("ex99|exhibit99",  case=False)]
    if matches.empty:
        print(f"No ex99 file in {accession_num}")
        return
    press_release_name = matches["name"].iloc[0]

    # Download the press release (uses cik_num, not apple_cik_num)
    file_url = f"https://www.sec.gov/Archives/edgar/data/{cik_num}/{accession}/{press_release_name}"
    file_response = requests.get(file_url, headers=my_dict)
    time.sleep(0.2)
    if file_response.status_code != 200:
        print(f"Failed file {accession_num}: {file_response.status_code}")
        return

    # Save the raw HTML
    
    os.makedirs(folder_path, exist_ok=True)
    
    with open(file_path, "w", encoding="utf-8") as f:
        f.write(file_response.text)

    print(f"Saved {file_path}")
```
