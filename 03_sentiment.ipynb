```python
import pandas as pd
import nltk
from  vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import re
```

```python
filings= pd.read_csv("data/processed/filings.csv",dtype={"accession": str})
print(filings.shape)
filings.head()
```

```python
sentiment_analysis=SentimentIntensityAnalyzer()
```

```python
test=filings["text"].iloc[0]
```

```python
filings["vader_compound"] =filings["text"].apply(lambda x:sentiment_analysis.polarity_scores(x)["compound"])
```

```python
filings_clean=filings[filings["length"]<=8000]
print(filings_clean.shape)
filings_clean.groupby("ticker").size()
```

```python
filings_clean["vader_compound"].describe()
```

```python
reference_documnet=pd.read_csv("data/reference/Loughran-McDonald_MasterDictionary_1993-2025.csv")
reference_documnet.head()
```

```python
reference_documnet.describe()
```

```python
negative_words = set(reference_documnet[reference_documnet["Negative"] > 0]["Word"])
positive_words = set(reference_documnet[reference_documnet["Positive"] > 0]["Word"])

print(len(negative_words), len(positive_words))
```

```python
def lm_score(text):
    words=re.findall(r"[a-z]+",text.lower())
    if len(words)==0:
        return 0
    pos=sum(1 for w in words if w.upper() in positive_words)
    neg=sum(1 for w in words if w.upper() in negative_words)
    return (pos-neg)/len(words)
```

```python
print(lm_score(filings_clean["text"].iloc[0]))
```

```python
filings_clean["lm_score"]=filings_clean["text"].apply(lm_score)
filings_clean["lm_score"].describe()
```

```python
filings_clean[["vader_compound", "lm_score"]].describe()
```

```python
for col in ["vader_compound", "lm_score"]:
    s = filings_clean[col]
    spread = (s.max() - s.min())
    print(col, "std:", round(s.std(), 4), "range:", round(spread, 4), "std/range:", round(s.std()/spread, 3))
```

```python
print("vader skew:", round(filings_clean["vader_compound"].skew(), 2))
print("lm skew:", round(filings_clean["lm_score"].skew(), 2))
```

```python
filings_clean.to_csv("data/processed/filings_scored.csv", index=False)
```
